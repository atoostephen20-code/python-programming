Variable names can be short (`a`, `x`, `y`, etc.) or descriptive ( `my_favorite_color`, `profit_margin`, `the_3_musketeers`, etc.). However, you must follow these rules while naming Python variables:

* A variable's name must start with a letter or the underscore character `_`. It cannot begin with a number.
* A variable name can only contain lowercase (small) or uppercase (capital) letters, digits, or underscores (`a`-`z`, `A`-`Z`, `0`-`9`, and `_`).
* Variable names are case-sensitive, i.e., `a_variable`, `A_Variable`, and `A_VARIABLE` are all different variables.

### Explanation: TypeError in String Multiplication

This cell produces a `TypeError` because of how the multiplication operator `*` works with strings:

1. **Replication**: In Python, you can multiply a **string** by an **integer** (e.g., `'Philip' * 3`). This results in the string being repeated that many times (`'PhilipPhilipPhilip'`).
2. **The Error**: You **cannot** multiply a string by another string (`name * name`). Python does not have a defined mathematical or logical operation for 'squaring' text.

**To fix this**: Ensure that the second operand is an integer if you intend to repeat the text, or use addition (`+`) if you intended to join (concatenate) strings together.

```python
name_5 = name * 5
name_5
```

#### String Methods

Strings in Python have many built-in *methods* that are used to manipulate them. Let's try out some common string methods.

### Tuple

A tuple is an ordered collection of values, similar to a list. However, it is not possible to add, remove, or modify values in a tuple. A tuple is created by enclosing values within parentheses `(` and `)`, separated by commas.

> Any data structure that cannot be modified after creation is called *immutable*. You can think of tuples as immutable lists.

Let's try some experiments with tuples.

```python
fruits = ('apple', 'cherry', 'dates', 'cherry')
fruits
```

```python
# check no. of elements
len(fruits)
```

```python
# get an element (negative index)
fruits[-2]
```

```python
# check if it contains an element
'dates' in fruits
```

```python
# TypeError: Tuples are immutable, meaning their elements cannot be changed after creation.
fruits[0] = 'avocado'
```

```python
# AttributeError: Tuples are immutable and do not have an `append()` method to add new elements.
fruits.append('blueberry')
```

```python
# To 'remove' an item from a tuple, convert it to a list first:
fruits_list = list(fruits)

# Check if 'apple' is in the list before trying to remove it
if 'apple' in fruits_list:
    fruits_list.remove('apple')
    print("'apple' removed.")
else:
    print("'apple' not found in the collection.")

fruits = tuple(fruits_list)
print(fruits)
```

```python
# AttributeError: Tuples are immutable and do not have a `pop()` method.
# To fix this, convert to a list, pop the element, and convert back to a tuple.
fruits_list = list(fruits)
if fruits_list:
    popped_item = fruits_list.pop()
    fruits = tuple(fruits_list)
    print(f"Popped item: {popped_item}")
    print(f"Updated tuple: {fruits}")
```

# 4. The if-else Statement
Syntax:
if condition:
    statement1
else:
    statement2

Example:

```python

age = 19

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

# Exercise

# 5. The if-elif-else Statement
Used when multiple conditions exist.

Example:

```python

score = 90

if score >= 70:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 50:
    print("Grade C")
else:
    print("Fail")
```

An if statement inside another if statement.

# 8. Introduction to Loops
Loops allow us to repeat tasks automatically.
Python has:
1.	for loop
2.	while loop



# The for Loop
Syntax:

for variable in sequence:
    statements



Example:

```python
for i in range(7):
    print(i)
```

# Understanding range()
* range(stop)
* range(start, stop)
* range(start, stop, step)


Examples:

```python

for i in range(1, 6):
    print(i)
```

# Counting Backward
Reversing the range


```python
for i in range(10, 0, -1):
    print(i)
```

# 11. The while Loop
Repeats as long as a condition remains True.
Syntax:
while condition:
    statements


Example:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

# 13. break Statement
Used to terminate a loop immediately.



Example:

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

# 14. Continue Statement
Skips the current iteration.


Example:

```python

for i in range(10):
    if i == 5:
        continue
    print(i)
```

# 15. Practical Examples
Example 1: Multiplication Table


```python
number = int(input("Enter a number: "))

for i in range(1, 13):
    print(number, "x", i, "=", number * i)
```

Example 2: Sum of Numbers

```python
total = 0

for i in range(1, 11):
    total += i

print("Sum =", total)
```

Example 3: Guessing Game

```python
secret = 7

guess = int(input("Guess the number: "))

while guess != secret:
    print("Try Again")
    guess = int(input("Guess again: "))

print("Correct!")
```
